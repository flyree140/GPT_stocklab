## 建立無未來洩漏快照
以下指令先建立 snapshot。只有把 `REVEAL` 改為 `True` 才會另外產生 results 檔。


```python
import os, subprocess
os.environ['ENABLE_HF_SENTIMENT'] = '1'
os.environ['ENABLE_QWEN'] = '1'
os.environ['QWEN_DAILY_LIMIT'] = '12'
cmd = ['python', 'scripts/backtest.py', '--as-of', AS_OF]
if REVEAL: cmd.append('--reveal')
subprocess.run(cmd, check=True)
subprocess.run(['python', 'scripts/quality_check.py'], check=True)
```

```python
from pathlib import Path
from IPython.display import FileLink
snapshot = Path('data/snapshots') / f'{AS_OF}.json'
display(FileLink(str(snapshot)))
if REVEAL:
    display(FileLink(str(Path('data/results') / f'{AS_OF}.json')) )
```
